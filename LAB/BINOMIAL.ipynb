```python
%%writefile binomial.c
#include <stdio.h>

int main()
{
    int n, k;
    int i, j;

    printf("Enter value of n: ");
    scanf("%d", &n);

    printf("Enter value of k: ");
    scanf("%d", &k);

    int C[n + 1][k + 1];

    for(i = 0; i <= n; i++)
    {
        for(j = 0; j <= k && j <= i; j++)
        {
            if(j == 0 || j == i)
                C[i][j] = 1;
            else
                C[i][j] = C[i - 1][j - 1] + C[i - 1][j];
        }
    }

    printf("\nBinomial Coefficient C(%d,%d) = %d\n", n, k, C[n][k]);

    return 0;
}
```
