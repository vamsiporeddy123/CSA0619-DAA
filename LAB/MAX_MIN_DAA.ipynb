```python
%%writefile maxmin.c
#include <stdio.h>

int max,min;

void maxMin(int a[], int low, int high)
{
    int mid,max1,min1;

    if(low==high)
    {
        max=min=a[low];
    }
    else if(high==low+1)
    {
        if(a[low]>a[high])
        {
            max=a[low];
            min=a[high];
        }
        else
        {
            max=a[high];
            min=a[low];
        }
    }
    else
    {
        mid=(low+high)/2;

        maxMin(a,low,mid);
        max1=max;
        min1=min;

        maxMin(a,mid+1,high);

        if(max1>max)
            max=max1;

        if(min1<min)
            min=min1;
    }
}

int main()
{
    int n,i,a[100];

    printf("Enter number of elements: ");
    scanf("%d",&n);

    printf("Enter %d elements:\n",n);
    for(i=0;i<n;i++)
        scanf("%d",&a[i]);

    maxMin(a,0,n-1);

    printf("\nMaximum Element = %d",max);
    printf("\nMinimum Element = %d\n",min);

    return 0;
}
```
