```python
%%writefile dice.c
#include <stdio.h>

int main()
{
    int dice, faces, sum;
    int i, j, k;

    printf("Enter number of dice: ");
    scanf("%d", &dice);

    printf("Enter number of faces: ");
    scanf("%d", &faces);

    printf("Enter required sum: ");
    scanf("%d", &sum);

    int dp[dice + 1][sum + 1];

    for(i = 0; i <= dice; i++)
        for(j = 0; j <= sum; j++)
            dp[i][j] = 0;

    dp[0][0] = 1;

    for(i = 1; i <= dice; i++)
    {
        for(j = 1; j <= sum; j++)
        {
            for(k = 1; k <= faces && k <= j; k++)
            {
                dp[i][j] += dp[i - 1][j - k];
            }
        }
    }

    printf("\nNumber of ways = %d\n", dp[dice][sum]);

    return 0;
}
```
