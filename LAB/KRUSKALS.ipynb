```python
# Kruskal's Minimum Spanning Tree

edges = [
    (0, 1, 10),
    (0, 2, 6),
    (0, 3, 5),
    (1, 3, 15),
    (2, 3, 4)
]

edges.sort(key=lambda x: x[2])

parent = [0, 1, 2, 3]

def find(x):
    while parent[x] != x:
        x = parent[x]
    return x

total = 0

for u, v, w in edges:
    a = find(u)
    b = find(v)

    if a != b:
        parent[a] = b
        total += w
        print(u, "-", v, "=", w)

print("Minimum Cost:", total)
```
