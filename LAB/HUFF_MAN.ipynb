```python
# Huffman Coding

import heapq

chars = ['a', 'b', 'c', 'd', 'e', 'f']
freq = [5, 9, 12, 13, 16, 45]

heap = [[freq[i], chars[i]] for i in range(len(chars))]
heapq.heapify(heap)

while len(heap) > 1:
    left = heapq.heappop(heap)
    right = heapq.heappop(heap)

    heapq.heappush(heap, [left[0] + right[0],
                          left[1] + right[1]])

print("Huffman Tree:", heap[0])
```
