```python
# Job Sequencing with Deadlines

jobs = [
    ('J1', 2, 100),
    ('J2', 1, 19),
    ('J3', 2, 27),
    ('J4', 1, 25),
    ('J5', 3, 15)
]

jobs.sort(key=lambda x: x[2], reverse=True)

slots = [None] * 3
profit = 0

for job, deadline, p in jobs:
    for j in range(min(deadline, 3) - 1, -1, -1):
        if slots[j] is None:
            slots[j] = job
            profit += p
            break

print("Jobs:", slots)
print("Maximum Profit:", profit)
```
