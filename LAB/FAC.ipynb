```python
%%writefile factorial.c
#include <stdio.h>

int main()
{
    int n, i;
    long long fact = 1;

    printf("Enter a number: ");
    scanf("%d", &n);

    for(i = 1; i <= n; i++)
        fact = fact * i;

    printf("\nFactorial of %d = %lld\n", n, fact);

    return 0;
}
```
