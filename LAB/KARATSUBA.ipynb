```python
%%writefile karatsuba.c
#include <stdio.h>
#include <math.h>

long long karatsuba(long long x, long long y)
{
    if (x < 10 || y < 10)
        return x * y;

    int n = (int)fmax(log10(x) + 1, log10(y) + 1);
    int m = n / 2;

    long long p = 1;
    int i;
    for(i = 0; i < m; i++)
        p *= 10;

    long long high1 = x / p;
    long long low1  = x % p;
    long long high2 = y / p;
    long long low2  = y % p;

    long long z0 = karatsuba(low1, low2);
    long long z1 = karatsuba(low1 + high1, low2 + high2);
    long long z2 = karatsuba(high1, high2);

    return z2 * p * p + (z1 - z2 - z0) * p + z0;
}

int main()
{
    long long a, b;

    printf("Enter first number: ");
    scanf("%lld", &a);

    printf("Enter second number: ");
    scanf("%lld", &b);

    printf("\nProduct = %lld\n", karatsuba(a, b));

    return 0;
}
```
