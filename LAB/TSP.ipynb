```python
# Travelling Salesman Problem

from itertools import permutations

graph = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

n = len(graph)
best = 9999

for path in permutations(range(1, n)):

    cost = graph[0][path[0]]

    for i in range(len(path)-1):
        cost += graph[path[i]][path[i+1]]

    cost += graph[path[-1]][0]

    best = min(best, cost)

print("Minimum TSP Cost:", best)
```
