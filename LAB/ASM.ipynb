```python
%%writefile asm.c
#include <stdio.h>

int main()
{
    int n;
    int a[2][20], t[2][20];
    int e[2], x[2];
    int f1[20], f2[20];
    int i, j;

    printf("Enter number of stations: ");
    scanf("%d", &n);

    printf("Enter processing time for Line 1:\n");
    for(i = 0; i < n; i++)
        scanf("%d", &a[0][i]);

    printf("Enter processing time for Line 2:\n");
    for(i = 0; i < n; i++)
        scanf("%d", &a[1][i]);

    printf("Enter transfer time from Line 1 to Line 2:\n");
    for(i = 1; i < n; i++)
        scanf("%d", &t[0][i]);

    printf("Enter transfer time from Line 2 to Line 1:\n");
    for(i = 1; i < n; i++)
        scanf("%d", &t[1][i]);

    printf("Enter entry times (e1 e2): ");
    scanf("%d %d", &e[0], &e[1]);

    printf("Enter exit times (x1 x2): ");
    scanf("%d %d", &x[0], &x[1]);

    f1[0] = e[0] + a[0][0];
    f2[0] = e[1] + a[1][0];

    for(i = 1; i < n; i++)
    {
        if(f1[i-1] + a[0][i] <= f2[i-1] + t[1][i] + a[0][i])
            f1[i] = f1[i-1] + a[0][i];
        else
            f1[i] = f2[i-1] + t[1][i] + a[0][i];

        if(f2[i-1] + a[1][i] <= f1[i-1] + t[0][i] + a[1][i])
            f2[i] = f2[i-1] + a[1][i];
        else
            f2[i] = f1[i-1] + t[0][i] + a[1][i];
    }

    if(f1[n-1] + x[0] <= f2[n-1] + x[1])
        printf("\nMinimum Time = %d\n", f1[n-1] + x[0]);
    else
        printf("\nMinimum Time = %d\n", f2[n-1] + x[1]);

    return 0;
}
```
