```python
# Bellman-Ford Algorithm

V = 5
edges = [
    (0, 1, 6),
    (0, 2, 7),
    (1, 2, 8),
    (1, 3, 5),
    (1, 4, -4),
    (2, 3, -3),
    (2, 4, 9),
    (3, 1, -2),
    (4, 0, 2),
    (4, 3, 7)
]

dist = [999] * V
dist[0] = 0

for _ in range(V - 1):
    for u, v, w in edges:
        if dist[u] != 999 and dist[u] + w < dist[v]:
            dist[v] = dist[u] + w

print("Shortest distances from source 0:")
print(dist)
```
