```python
# Warshall Algorithm - Transitive Closure

graph = [
    [1, 1, 0],
    [0, 1, 1],
    [0, 0, 1]
]

n = len(graph)

for k in range(n):
    for i in range(n):
        for j in range(n):
            graph[i][j] = graph[i][j] or (
                graph[i][k] and graph[k][j]
            )

print("Transitive Closure:")

for row in graph:
    print(row)
```
