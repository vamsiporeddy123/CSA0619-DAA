```python
# Boruvka's Minimum Spanning Tree

edges = [
    (0, 1, 10),
    (0, 2, 6),
    (0, 3, 5),
    (1, 3, 15),
    (2, 3, 4)
]

n = 4
parent = list(range(n))

def find(x):
    if parent[x] != x:
        parent[x] = find(parent[x])
    return parent[x]

def union(a, b):
    a = find(a)
    b = find(b)
    if a != b:
        parent[b] = a
        return True
    return False

components = n
total = 0

while components > 1:

    cheapest = [None] * n

    for u, v, w in edges:
        a = find(u)
        b = find(v)

        if a != b:
            if cheapest[a] is None or w < cheapest[a][2]:
                cheapest[a] = (u, v, w)

            if cheapest[b] is None or w < cheapest[b][2]:
                cheapest[b] = (u, v, w)

    for edge in cheapest:
        if edge:
            u, v, w = edge

            if union(u, v):
                print(u, "-", v, "=", w)
                total += w
                components -= 1

print("Minimum Cost:", total)
```
