```python
%%writefile mom.c
#include <stdio.h>
#include <stdlib.h>

int compare(const void *a, const void *b)
{
    return (*(int *)a - *(int *)b);
}

int main()
{
    int n, k, i;
    int a[100];

    printf("Enter number of elements: ");
    scanf("%d", &n);

    printf("Enter %d elements:\n", n);
    for(i = 0; i < n; i++)
        scanf("%d", &a[i]);

    printf("Enter value of K: ");
    scanf("%d", &k);

    qsort(a, n, sizeof(int), compare);

    printf("\n%dth Smallest Element = %d\n", k, a[k-1]);

    return 0;
}
```
