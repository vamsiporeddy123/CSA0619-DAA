```python
# Prim's Minimum Spanning Tree

INF = 999

graph = [
    [0, 2, 0, 6, 0],
    [2, 0, 3, 8, 5],
    [0, 3, 0, 0, 7],
    [6, 8, 0, 0, 9],
    [0, 5, 7, 9, 0]
]

n = len(graph)
selected = [False] * n
key = [INF] * n

key[0] = 0
total = 0

for _ in range(n):
    u = -1

    for i in range(n):
        if not selected[i] and (u == -1 or key[i] < key[u]):
            u = i

    selected[u] = True
    total += key[u]

    for v in range(n):
        if graph[u][v] and not selected[v]:
            key[v] = min(key[v], graph[u][v])

print("Minimum Cost:", total)
```
